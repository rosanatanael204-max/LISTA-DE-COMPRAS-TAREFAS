Funcionamento do DBSCAN e Diferença para o K-MeansO DBSCAN (Density-Based Spatial Clustering of Applications with Noise) é um algoritmo de agrupamento baseado em densidade.Como funciona: Ele agrupa pontos que estão próximos uns dos outros (com base em um raio de vizinhança $\epsilon$ - eps e em um número mínimo de pontos min_samples). Pontos em áreas de baixa densidade são automaticamente classificados como ruído (noise).Principais Vantagens: Não exige que você defina o número de clusters (k) previamente e consegue identificar clusters de formatos arbitrários (não apenas esféricos), além de lidar muito bem com ruídos e outliers.Diferença para o K-Means:Número de Clusters: O K-Means exige que o usuário defina k de antemão; o DBSCAN descobre o número de clusters automaticamente com base na densidade dos dados.Formato: O K-Means assume clusters convexos e esféricos e tende a dividir os dados em tamanhos iguais; o DBSCAN identifica formas complexas e arbitrárias.Ruído: O K-Means força todos os pontos a pertencerem a algum cluster; o DBSCAN identifica pontos isolados como ruído (-1), evitando que distorçam os agrupamentos.

Descrição da Base de Dados
A base de dados utilizada é a Breast Cancer Wisconsin (Diagnostic), disponível nativamente no scikit-learn.

Origem: Contém características computadas a partir de imagens digitalizadas de punções aspirativas por agulha fina (PAAF) de massas mamárias.

Atributos: Possui 30 características numéricas contínuas (como raio, textura, perímetro, área, suavidade, compacidade, concavidade, etc., nas formas média, erro e pior valor).

Amostras: 569 instâncias no total.

Classes (target): 2 classes conhecidas (diagnóstico):

0: Maligno (Malignant) - 212 amostras

1: Benigno (Benign) - 357 amostras
(Nota: Embora o problema original seja de classificação supervisionada, utilizamos a base aqui para tarefas não supervisionadas de agrupamento, comparando depois os clusters formados com as classes reais através de métricas).

3. Interpretação das Métricas Escolhidas
Para avaliar a qualidade dos clusters gerados, utilizamos duas métricas principais:

Índice de Silhueta (Silhouette Score):

Mede quão próximo cada ponto em um cluster está dos pontos dos clusters vizinhos. Varia de -1 a +1.

Interpretação: Valores próximos de +1 indicam que os clusters estão bem separados e densos. Valores próximos de 0 indicam sobreposição, e valores negativos indicam que os pontos podem ter sido colocados no cluster errado. (Nota: Pontos de ruído do DBSCAN são desconsiderados no cálculo da silhueta padrão).

Adjusted Rand Index (ARI):

Como temos os rótulos reais (y original) da base de dados, o ARI mede a similaridade entre os clusters gerados pelo DBSCAN e as classes reais, ajustando para o acaso. Varia de -1 a 1 (onde 1 significa concordância perfeita).

Interpretação: Um ARI próximo de 0 indica agrupamento aleatório em relação às classes reais, enquanto valores positivos indicam concordância significativa entre a estrutura de densidade descoberta e o diagnóstico médico real.

Importação de Bibliotecas e Carregamento dos Dados

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA

# Carregando a base de dados Breast Cancer
cancer = load_breast_cancer()
X = cancer.data
y = cancer.target
feature_names = cancer.feature_names

print(f"Dimensões da base de dados: {X.shape}")
print(f"Classes (0: Maligno, 1: Benigno): {np.bincount(y)}")
```

Pré-processamento (Normalização)
O DBSCAN é baseado em distâncias euclidianas, portanto a normalização (padronização) dos dados é obrigatória para que todas as 30 features tenham o mesmo peso.

```python
# Padronizando as features para média 0 e desvio padrão 1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Aplicação do DBSCAN e Redução de Dimensionalidade (PCA)
Como a base possui 30 dimensões, aplicaremos o PCA apenas para reduzir para 2 componentes principais com a finalidade de visualização gráfica em 2D, mantendo a essência da variância dos dados.

```python
# Aplicando DBSCAN nos dados normalizados
# Ajuste de eps e min_samples para encontrar um equilíbrio entre clusters e ruído
dbscan = DBSCAN(eps=3.5, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)

# Redução de dimensionalidade para visualização 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# DataFrame para facilitar a plotagem
df_vis = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_vis['Cluster'] = clusters
df_vis['Target_Real'] = y
```

Cálculo das Métricas de Avaliação

```python
# 1. Índice de Silhueta (excluindo os pontos de ruído -1)
valid_mask = clusters != -1
n_clusters_valid = len(np.unique(clusters[valid_mask]))

if n_clusters_valid > 1:
    sil_score = silhouette_score(X_scaled[valid_mask], clusters[valid_mask])
else:
    sil_score = -1.0

# 2. Adjusted Rand Index (comparando com as classes reais y)
ari_score = adjusted_rand_score(y, clusters)

# Estatísticas gerais
n_noise = list(clusters).count(-1)
total_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)

print(f"=== RESULTADOS DA AVALIAÇÃO ===")
print(f"Número de clusters encontrados: {total_clusters}")
print(f"Número de pontos de ruído (Outliers): {n_noise}")
print(f"Índice de Silhueta (sem ruído): {sil_score:.4f}")
print(f"Adjusted Rand Index (ARI): {ari_score:.4f}")
```

Visualização Gráfica dos Clusters

```python
plt.figure(figsize=(12, 5))

# Gráfico 1: Clusters gerados pelo DBSCAN
plt.subplot(1, 2, 1)
sns.scatterplot(
    x='PC1', y='PC2',
    hue='Cluster',
    palette='Set1',
    data=df_vis,
    alpha=0.8,
    s=50,
    legend='full'
)
plt.title('Clusters gerados pelo DBSCAN (Ruído = -1)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')

# Gráfico 2: Classes reais (Maligno vs Benigno) para comparação
plt.subplot(1, 2, 2)
sns.scatterplot(
    x='PC1', y='PC2',
    hue='Target_Real',
    palette='coolwarm',
    data=df_vis,
    alpha=0.8,
    s=50
)
plt.title('Classes Reais (0: Maligno, 1: Benigno)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')

plt.tight_layout()
plt.show()
```

Análise dos Resultados e Visualização
Interpretação da Visualização: No gráfico gerado pelo DBSCAN, você observará que o algoritmo agrupa a grande massa densa de dados em um cluster principal e isola pontos periféricos como ruído (-1). Ao comparar com o gráfico das classes reais ao lado, nota-se que a separação linear clássica dos diagnósticos de câncer de mama é complexa para algoritmos puramente baseados em densidade de alta dimensão sem ajustes finos, evidenciando o desafio de aplicar aprendizado não supervisionado em dados médicos altamente correlacionados.

Interpretação das Métricas: O Índice de Silhueta obtido reflete a coesão dos pontos nos agrupamentos densos formados, enquanto o ARI mostra o grau de alinhamento entre a separação por densidade do DBSCAN e o gabarito clínico real
